## Section 3, build the model

```python
from sklearn.linear_model import LinearRegression
```

```python
# https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.LinearRegression.html
LinearRegression?
```

Create the model, with sklearn you initialise the model with an "empty constructor" of the base form of the model. LinearRegression is the constructor

```python
model = LinearRegression()
```

Now "fit" the model using X and y. Use an appropriate ? after method name to see. .fit expects the X part to be a matrix, but we have a 1 dimensional array. Pay attention to the error message when you just do x,y

```python
model.fit(X,y)
```

## Section 4 Inferences

```python
model.coef_
```

```python
model.intercept_
```

There are the parameters

$\hat{y} = 0.2142857142857153 + 2.12857143 x$

is the model

```python
w0 = model.intercept_
w1 = model.coef_[0]
```

```python
predictions = w0 + np.dot(w1,X).reshape(1,-1)
```

```python
predictions
```

Using the built in .predict

```python
pred = model.predict(X)
```

```python
pred
```

When to use MSE: 
- mean difference between value and prediction 
- it penalizes more big errors 
- use it to optmise models 
- sometimes root MSE is even better, bc it gives you a value in the same scale as your dataset values 

when to use r2 score: 
- it tells you on a scale 0-1 how much the model can explain value variability. Close to 1 is good. 
- normalized (percentage). Useful to compare models 
- BUT it gave give weight to irrelevant params and induce misunderstanding. 


Very close to 1!

```python
rms = rms(y,pred)
rms
```
